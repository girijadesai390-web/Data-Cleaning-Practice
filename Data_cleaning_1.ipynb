```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
data = {
    "Name":[" John","Alice","johnny","Bob","priya","Alice",None],
    "Age":[24,23,25,None,25,23,35],
    "City":["Pune","sangli","Kolhapur ","Mumbai"," kolhapur","sangli",None],
    "Gender":["Male","female","M","male","f","female","M"]
}

df = pd.DataFrame(data)
df
```

```python
df.isnull().sum()
```

```python
df.describe()
```

```python
df["Age"].fillna(df["Age"].mean(),inplace=True)
df
```

```python
df.isnull().sum()
```

```python
df["Name"].fillna("unknown",inplace=True)
df
```

```python
df["City"].fillna("unknown",inplace=True)
df
```

```python
df.isnull().sum()
```

```python
df["Name"]=df["Name"].str.title()
df
```

```python
df["City"]=df["City"].str.capitalize()
df
```

```python
df["City"]=df["City"].str.strip()
df["City"]=df["City"].str.capitalize()
df
```

```python
df["Name"]=df["Name"].str.strip()
df
```

```python
def clean_gender(gender):
  if gender=="M" or gender=="Male" or gender=="male":
    return "Male"
  elif gender=="f" or gender=="female" or gender=="Female":
    return "Female"
  else:
    return "Unknown"
```

```python
df["Gender"] =df["Gender"].apply(clean_gender)
df
```

```python
df=df.drop_duplicates()
df
```

```python
df.info()
```

```python
np.dtype(df["Age"])
```

```python
df["Age"] = df["Age"].astype(int)
df.info()
```
